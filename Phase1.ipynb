# Phase 1: Data Selection
### Telco Customer Churn — Data Mining Project (IT326)

## 1. Project Goal

The purpose of this project is to apply data mining techniques to the Telco Customer Churn dataset in order to:

- **Classification:** predict whether a customer will churn (`Churn` = Yes/No) based on their demographic, account, and service attributes.
- **Clustering:** group customers into segments that share similar characteristics and usage behavior, without using the churn label.

Understanding and predicting churn helps a telecom company identify at-risk customers early and take proactive retention action, which is far cheaper than acquiring new customers.

## 2. Dataset Source

- **Name:** Telco Customer Churn
- **Source / URL:** [Kaggle — Telco Customer Churn](https://www.kaggle.com/datasets/blastchar/telco-customer-churn) (originally published by IBM as a sample dataset)
- **Local copy used in this project:** [`Dataset/Raw_dataset.csv`](Dataset/Raw_dataset.csv)
- **Related research paper:** Ahmad, A.K., Jafar, A. & Aljoumaa, K. "Customer churn prediction in telecom using machine learning in big data platform." *J Big Data* 6, 28 (2019). https://doi.org/10.1186/s40537-019-0191-6 (see [`Dataset/Ahmad_et_al_2019_Customer_Churn_Prediction_Telecom_BigData.pdf`](Dataset/Ahmad_et_al_2019_Customer_Churn_Prediction_Telecom_BigData.pdf))

### Class attribute: `Churn`

The target class attribute is `Churn`, with two possible values: `Yes` (customer left) and `No` (customer stayed).

```python
class_counts = df['Churn'].value_counts()
class_percent = df['Churn'].value_counts(normalize=True) * 100

summary = pd.DataFrame({'Count': class_counts, 'Percentage': class_percent.round(2)})
summary
```
